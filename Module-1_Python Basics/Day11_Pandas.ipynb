```python
# Binning:
df2
# if person is married, graduate and not self employed ---> stable
# else --> unstable
```

```python
def bin(x):
    if (x["Married"]=="Yes")& (x["Education"]=="Graduate") & (x["Self_Employed"]=="No"):
        return "Stable"
    else:
        return"Unstable"
```

```python
df2["Bin"]=df2.apply(bin,axis=1)
```
