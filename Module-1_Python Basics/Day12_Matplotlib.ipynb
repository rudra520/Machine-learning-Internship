```python
# Bar Plot
x_1=[5,3,4,6,2]
y_1=[10,20,34,23,21]

plt.bar(x_1,y_1,color=["red","green","blue","pink","brown"])

plt.xlabel("Sales")
plt.ylabel("Profit")

plt.title("Sales portfolio")
```

```python
# Histogram

axis_1 = [10,20,3,45]

plt.hist(y)
```

```python
# Scatterplot

x=[10,20,30,40]
y=[10,5,8,4]

plt.scatter(x,y)
```

```python
# Stacked plots
var_1=["A","B","C","D"]
y1=[10,20,30,10]
y2=[20,25,15,25]

plt.bar(var_1,y2,color="cyan")
plt.bar(var_1,y1)
```

```python
# pie chart
label_1=["MEN","Women","Other"]
x=[20,30,40]

plt.title("PIE")
plt.pie(x,labels=label_1,explode=[0.5,0.2])
plt.legend()
```
